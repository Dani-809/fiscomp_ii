Ejercicio 1

```python
import numpy as np
```

```python
rng = np.random.default_rng( seed= 8 )
```

```python
L_list = [1.0, 2.5, 0.8, 5.0, 10.0]
d_list = [10, 25, 15, 40, 60]
```

```python
L = np.array(L_list)
d = np.array(d_list)
```

```python
L_sol = 3.828e26
PC = 3.0857e16
```

```python
def flujo_estrella(L,d) :
  Lum = L * L_sol
  Mts = d * PC
  return Lum / (4 * np.pi * Mts**2)
```

```python
flujo = flujo_estrella(L,d)
```

```python
indice_max = np.argmax(flujo)
indice_min = np.argmin(flujo)
print(indice_max)
print(indice_min)
```

```python
prom_flujo = np.mean(flujo)
med_flujo = np.median(flujo)

print(prom_flujo)
print(med_flujo)
```

```python
lim = 1e-11
mascara = flujo >= lim
print(mascara)
```

```python
indices_detectadas = np.where(mascara)
print(indices_detectadas)
```

```python
indice_ordenado = np.argsort(flujo)[::-1]
print(indice_ordenado)
```

Ejercicio 2

```python
t = np.linspace(0, 10, 500)
flujo = np.ones(500)
```

```python
transito = (4.5 <= t) & (t <= 5.0)
flujo[transito] -= 0.02
```

```python
ruido_g = rng.normal(loc=0, scale=0.002, size=(500,))
f_ruidoso = flujo + ruido_g
```

```python
flujo_prom = np.mean(f_ruidoso)
print("flujo promedio:", flujo_prom)
```

```python
sigma = np.std(f_ruidoso)
print("desviacion estandar:", sigma)
```

```python
min = np.argmin(f_ruidoso)
flujo_min = f_ruidoso[min]
print("flujo minimo:", flujo_min)
print("indice:", min)
```

```python
tiempo_min = t[min]
print(f"tiempo minimo: {tiempo_min} años")
```

```python
#e)
mascara = f_ruidoso < (flujo_prom - 3 * sigma)
puntos = f_ruidoso[mascara]
```

```python
#f
cant = np.sum(mascara)
print("cantidad de puntos identificados:", cant)
```

```python
#g
F_transito = np.mean(puntos)
delta_F = 1 - F_transito
print("profundidad del transito:", delta_F)
```

```python
r_radios = np.sqrt(delta_F)
print("razon entre radios (Rp/R*):", r_radios)
```

```python
tiempos_identificados = t[mascara]
print("intervalo del transito: [4.5, 5.0] años")
print("inicio detectado:", tiempos_identificados.min())
print("final detectado:", tiempos_identificados.max())
```
